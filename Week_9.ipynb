```python
import pandas as pd

# Load the dataset
df = pd.read_csv('/content/hospital_patients_real_world.csv')

# Display the first 5 rows of the DataFrame
print("First 5 rows of the dataset:")
display(df.head())

# Display general information about the DataFrame, including data types and non-null values
print("\nDataFrame Information:")
df.info()
```

### Handle Missing Data

Let's detect missing values and then apply a basic strategy to handle them. For simplicity, we'll drop rows that contain any missing values. In a real-world scenario, you might choose to impute missing values (e.g., with the mean, median, or mode) depending on the nature of the data and the column.

```python
# Detect missing values
print("\nNumber of missing values per column (before handling):")
display(df.isnull().sum())

# Handle missing values: Drop rows with any missing values
df_cleaned = df.dropna()

print("\nNumber of missing values per column (after dropping rows with any NaNs):")
display(df_cleaned.isnull().sum())

print(f"\nOriginal number of rows: {len(df)}")
print(f"Number of rows after dropping NaNs: {len(df_cleaned)}")

display(df_cleaned.head())
```

### Detect and Filter Outliers

For detecting outliers, a common simple method is to use the Interquartile Range (IQR). Values that fall below `Q1 - 1.5 * IQR` or above `Q3 + 1.5 * IQR` are considered outliers. We'll apply this to a numerical column, for example, 'Age' or any relevant numerical column available in your dataset. If the dataset does not have suitable numerical columns for outlier detection, you may need to adjust the column selection.

Let's assume 'Age' is a numerical column suitable for outlier detection. If not, please replace 'Age' with an appropriate numerical column name from your dataset (e.g., 'Pulse Rate', 'Weight', etc.).

```python
# Identify a numerical column for outlier detection (e.g., 'Age')
# Please adjust 'numerical_column' if 'Age' is not suitable or doesn't exist
numerical_column = 'Age' # Or 'Pulse Rate', 'Weight', etc.

# Check if the column exists and is numeric
if numerical_column in df_cleaned.columns and pd.api.types.is_numeric_dtype(df_cleaned[numerical_column]):
    print(f"\nDetecting outliers in the '{numerical_column}' column using IQR method...")

    Q1 = df_cleaned[numerical_column].quantile(0.25)
    Q3 = df_cleaned[numerical_column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df_cleaned[(df_cleaned[numerical_column] < lower_bound) | (df_cleaned[numerical_column] > upper_bound)]
    df_no_outliers = df_cleaned[(df_cleaned[numerical_column] >= lower_bound) & (df_cleaned[numerical_column] <= upper_bound)]

    print(f"Number of outliers detected in '{numerical_column}': {len(outliers)}")
    print(f"Original number of rows in cleaned data: {len(df_cleaned)}")
    print(f"Number of rows after filtering outliers: {len(df_no_outliers)}")

    print(f"\nOutliers in '{numerical_column}':")
    display(outliers.head())

    print(f"\nDataFrame after removing outliers in '{numerical_column}':")
    display(df_no_outliers.head())

else:
    print(f"Column '{numerical_column}' not found or is not numeric. Please choose a valid numerical column for outlier detection.")
    print("Available columns:", df_cleaned.columns.tolist())
```

### 1. Load Data and Initial Overview (Simple)

```python
import pandas as pd

df_simple = pd.read_csv('/content/hospital_patients_real_world.csv')

print("First 5 rows:")
display(df_simple.head())

print("\nMissing values per column (count):")
display(df_simple.isnull().sum())
```

### 2. Handle Missing Data (Simple)

This code drops any row that contains at least one missing value. This is a quick way to clean the data, but for some datasets, imputation (filling missing values with estimates) might be more suitable.

```python
# Drop rows with any missing values
df_cleaned_simple = df_simple.dropna()

print(f"Original rows: {len(df_simple)}")
print(f"Rows after dropping NaNs: {len(df_cleaned_simple)}")

print("\nMissing values after cleaning:")
display(df_cleaned_simple.isnull().sum())
```

### 3. Detect and Filter Outliers (Simple)

We'll use the Interquartile Range (IQR) method for a numerical column. This method defines outliers as values significantly outside the 'middle 50%' of the data. For this example, we'll use the 'Age' column. If 'Age' is not present or not numeric, please update the `numerical_column` variable to another suitable column.

```python
numerical_column_simple = 'Age' # Change if needed

if numerical_column_simple in df_cleaned_simple.columns and pd.api.types.is_numeric_dtype(df_cleaned_simple[numerical_column_simple]):
    Q1 = df_cleaned_simple[numerical_column_simple].quantile(0.25)
    Q3 = df_cleaned_simple[numerical_column_simple].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Identify outliers
    outliers_mask = (df_cleaned_simple[numerical_column_simple] < lower_bound) | \
                    (df_cleaned_simple[numerical_column_simple] > upper_bound)
    outliers_count = outliers_mask.sum()

    # Filter out outliers
    df_no_outliers_simple = df_cleaned_simple[~outliers_mask]

    print(f"Outliers in '{numerical_column_simple}': {outliers_count} rows")
    print(f"Data size after removing outliers: {len(df_no_outliers_simple)} rows")

    print(f"\nFirst 5 rows after removing outliers in '{numerical_column_simple}':")
    display(df_no_outliers_simple.head())
else:
    print(f"Column '{numerical_column_simple}' not found or is not numeric. Available columns: {df_cleaned_simple.columns.tolist()}")
```
