```python
import pandas as pd

# Load the CSV file into a pandas DataFrame
df = pd.read_csv('/content/Simple 100 Student Marks.csv')
```

```python
# Display the first 5 rows of the DataFrame
display(df.head())
```

```python
# Display descriptive statistics for numerical columns
display(df.describe())
```

```python
# Calculate and display the mean marks for each subject
mean_marks = df[['maths_marks', 'science_marks', 'english_marks', 'social_studies_marks', 'language_marks']].mean()
print("Mean Marks for Each Subject:\n")
display(mean_marks)
```

```python
# Calculate and display the median marks for each subject
median_marks = df[['maths_marks', 'science_marks', 'english_marks', 'social_studies_marks', 'language_marks']].median()
print("Median Marks for Each Subject:\n")
display(median_marks)
```

```python
# Calculate and display the standard deviation of marks for each subject
std_dev_marks = df[['maths_marks', 'science_marks', 'english_marks', 'social_studies_marks', 'language_marks']].std()
print("Standard Deviation of Marks for Each Subject:\n")
display(std_dev_marks)
```

```python
# Calculate and display the minimum marks for each subject
min_marks = df[['maths_marks', 'science_marks', 'english_marks', 'social_studies_marks', 'language_marks']].min()
print("Minimum Marks for Each Subject:\n")
display(min_marks)
```

```python
# Calculate and display the maximum marks for each subject
max_marks = df[['maths_marks', 'science_marks', 'english_marks', 'social_studies_marks', 'language_marks']].max()
print("Maximum Marks for Each Subject:\n")
display(max_marks)
```

```python
# Identify students who scored above 80 in Maths
high_maths_scores = df[df['maths_marks'] > 80]
print("Students with Maths marks > 80:\n")
display(high_maths_scores[['student_name', 'maths_marks']].sort_values(by='maths_marks', ascending=False))
```

```python
# Calculate total marks for each student
df['total_marks'] = df[['maths_marks', 'science_marks', 'english_marks', 'social_studies_marks', 'language_marks']].sum(axis=1)

# Find the student with the highest total marks
top_student = df.loc[df['total_marks'].idxmax()]

print("Student with the highest total marks:\n")
display(top_student[['student_name', 'total_marks']])
```

```python
# Find the student with the lowest total marks
lowest_student = df.loc[df['total_marks'].idxmin()]

print("Student with the lowest total marks:\n")
display(lowest_student[['student_name', 'total_marks']])
```
