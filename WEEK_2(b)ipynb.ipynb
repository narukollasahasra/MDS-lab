```python
from PIL import Image
import numpy as np

# Open the image file
image = Image.open(image_filename)

# Convert the image to a NumPy array
image_array = np.array(image)

print(f"Image successfully converted to a NumPy array.\nShape of the array: {image_array.shape}")
```

```python
# Display the data type of the NumPy array
print(f"Data type of the image array: {image_array.dtype}")
```

```python
from PIL import Image
import numpy as np

# Reload the image, explicitly converting it to RGB mode to ensure color channels
color_image = Image.open(image_filename).convert('RGB')

# Convert the color image to a NumPy array
color_image_array = np.array(color_image)

print(f"Color image successfully loaded. Shape of the array: {color_image_array.shape}")
```

```python
from PIL import Image

# Adjusting coordinates to better capture only the eyes based on previous interaction
y_start = 300  # Starting row (from top)
y_end = 600   # Ending row
x_start = 950  # Starting column (from left)
x_end = 1600   # Ending column

# Crop the color image array
eyes_cropped_color_array = color_image_array[y_start:y_end, x_start:x_end, :]

# Convert the cropped NumPy array back to a PIL Image for display
eyes_cropped_color_image = Image.fromarray(eyes_cropped_color_array)

print(f"Cropped color image shape: {eyes_cropped_color_array.shape}")

# Display the cropped color image
display(eyes_cropped_color_image)
```

```python
import matplotlib.pyplot as plt
from PIL import Image

# Get the complete color image to perform transformations on
img_to_transform = color_image

# Perform transformations
img_flipped_horizontal = img_to_transform.transpose(Image.FLIP_LEFT_RIGHT)
img_flipped_vertical = img_to_transform.transpose(Image.FLIP_TOP_BOTTOM)
img_rotated_90 = img_to_transform.transpose(Image.ROTATE_90)
img_rotated_270 = img_to_transform.transpose(Image.ROTATE_270)

# Create a figure and a set of subplots
fig, axes = plt.subplots(2, 3, figsize=(15, 10))

# Display the original image
axes[0, 0].imshow(img_to_transform)
axes[0, 0].set_title('Original Image')
axes[0, 0].axis('off')

# Display the horizontally flipped image
axes[0, 1].imshow(img_flipped_horizontal)
axes[0, 1].set_title('Horizontal Flip')
axes[0, 1].axis('off')

# Display the vertically flipped image
axes[0, 2].imshow(img_flipped_vertical)
axes[0, 2].set_title('Vertical Flip')
axes[0, 2].axis('off')

# Display the 90-degree rotated image
axes[1, 0].imshow(img_rotated_90)
axes[1, 0].set_title('90 Degree Rotation')
axes[1, 0].axis('off')

# Display the 270-degree rotated image
axes[1, 1].imshow(img_rotated_270)
axes[1, 1].set_title('270 Degree Rotation')
axes[1, 1].axis('off')

# Hide the empty subplot
axes[1, 2].axis('off')

plt.tight_layout()
plt.show()
```
