```python
import time
import numpy as np

# Define the size of the lists/arrays for performance comparison
size = 10**6

# --- Python List comparison ---
# Record start time for Python list operations
start_time = time.time()
# Create two Python lists
py_list1 = list(range(size))
py_list2 = list(range(size))
# Perform element-wise addition using a list comprehension
py_result = [py_list1[i] + py_list2[i] for i in range(size)]
# Record end time and calculate duration for Python list
end_time = time.time()
python_list_time = end_time - start_time
print(f"Time taken for Python List operation: {python_list_time:.6f} seconds")

# --- NumPy Array comparison ---
# Record start time for NumPy array operations
start_time = time.time()
# Create two NumPy arrays
numpy_array1 = np.arange(size)
numpy_array2 = np.arange(size)
# Perform element-wise addition using NumPy's vectorized operation
numpy_result = numpy_array1 + numpy_array2
# Record end time and calculate duration for NumPy array
end_time = time.time()
numpy_array_time = end_time - start_time
print(f"Time taken for NumPy Array operation: {numpy_array_time:.6f} seconds")

# Compare the performance of NumPy versus Python lists
print(f"\nNumPy is approximately {python_list_time / numpy_array_time:.2f} times faster than Python list for this operation.")
```

```python
import sys

# Calculate memory usage for Python list
# sys.getsizeof() gets the size of the list object itself, then sum of sizes of each element
memory_py_list = sys.getsizeof(py_list1) + sum(sys.getsizeof(item) for item in py_list1)
print(f"Memory taken by Python List: {memory_py_list / (1024**2):.4f} MB")

# Calculate memory usage for NumPy array
# .nbytes gives the total bytes consumed by the array elements
memory_numpy_array = numpy_array1.nbytes
print(f"Memory taken by NumPy Array: {memory_numpy_array / (1024**2):.4f} MB")

# Compare memory usage between Python list and NumPy array
print(f"\nPython List uses approximately {memory_py_list / memory_numpy_array:.2f} times more memory than NumPy array for this data.")
```

## Summary of Time and Memory Comparison

Based on the operations performed:

### Time Performance
*   **Python List**: `0.1500` seconds
*   **NumPy Array**: `0.0083` seconds

NumPy Arrays were approximately **`18.14` times faster** than Python lists for the element-wise addition operation on `10^6` elements.

### Memory Usage
*   **Python List**: `34.3323` MB
*   **NumPy Array**: `7.6294` MB

Python Lists used approximately **`4.50` times more memory** than NumPy arrays for storing `10^6` integer elements. This is because Python lists store references to objects, while NumPy arrays store contiguous blocks of homogeneous data.
