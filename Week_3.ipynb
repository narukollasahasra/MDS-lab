```python
import pandas as pd

# Load the dataset
df = pd.read_csv('/content/Most Streamed Artists on Spotify (17_07_2026) V1.1.csv')

print("First 10 records:")
display(df.head(10))
```

```python
print("Last 10 records:")
display(df.tail(10))
```

```python
print("DataFrame Shape:", df.shape)
print("\nDataFrame Index:", df.index)
print("\nDataFrame Columns:", df.columns)
```

## Selecting Rows based on Conditions

We can select specific rows (records) from the DataFrame by applying a boolean condition. For example, let's filter the DataFrame to show only artists who are 'Male'.

```python
# Select artists where 'Sex' is 'Male'
male_artists_df = df[df['Sex'] == 'Male']

print("Number of Male Artists:", len(male_artists_df))
display(male_artists_df.head())
```

## Deleting Rows based on Conditions

To delete rows, we can select all rows that *do not* meet a certain condition. For example, let's create a new DataFrame that excludes artists who debuted after 2010.

```python
# Delete artists who debuted after 2010
artists_debuted_before_2010_df = df[df['Debut Year'] <= 2010]

print("Original number of artists:", len(df))
print("Number of artists debuted before or in 2010:", len(artists_debuted_before_2010_df))
display(artists_debuted_before_2010_df.head())
```

## Selecting Columns

We can select specific columns from the DataFrame by passing a list of column names. Let's create a new DataFrame with only the 'Artist Name', 'Primary Genre', and 'Total Streams (in millions)' columns.

```python
# Select specific columns
selected_columns_df = df[['Artist Name', 'Primary Genre', 'Total Streams (in millions)']]

print("Shape of DataFrame with selected columns:", selected_columns_df.shape)
display(selected_columns_df.head())
```

## Deleting Columns

To delete columns, we can use the `drop()` method, specifying the column name(s) and `axis=1` to indicate that we are dropping columns. Let's remove the 'Collaborative Streams (in millions)' column from the DataFrame.

```python
# Delete a specific column
df_columns_dropped = df.drop(columns=['Collaborative Streams (in millions)'])

print("Shape of DataFrame after dropping a column:", df_columns_dropped.shape)
display(df_columns_dropped.head())
```

## Sorting Data

We can sort the DataFrame by one or more columns using the `sort_values()` method. Let's sort the original DataFrame by 'Total Streams (in millions)' in descending order.

```python
# Sort the DataFrame by 'Total Streams (in millions)' in descending order
df_sorted_streams = df.sort_values(by='Total Streams (in millions)', ascending=False)

print("DataFrame sorted by Total Streams (top 5):")
display(df_sorted_streams.head())
```

## Ranking Data

We can assign ranks to the artists based on their 'Total Streams (in millions)'. Let's add a 'Rank' column to the sorted DataFrame.

```python
# Add a 'Rank' column based on 'Total Streams (in millions)'
df_sorted_streams['Rank'] = df_sorted_streams['Total Streams (in millions)'].rank(ascending=False, method='min')

print("DataFrame with 'Rank' column (top 5):")
display(df_sorted_streams.head())
```

## Basic Statistical Operations

Now, let's perform some basic statistical operations on the 'Total Streams (in millions)' column, specifically calculating its mean and median.

```python
# Calculate the mean of 'Total Streams (in millions)'
mean_total_streams = df['Total Streams (in millions)'].mean()

# Calculate the median of 'Total Streams (in millions)'
median_total_streams = df['Total Streams (in millions)'].median()

print(f"Mean of Total Streams (in millions): {mean_total_streams:.2f}")
print(f"Median of Total Streams (in millions): {median_total_streams:.2f}")
```

```python
# Calculate the standard deviation of 'Total Streams (in millions)'
std_total_streams = df['Total Streams (in millions)'].std()

print(f"Standard Deviation of Total Streams (in millions): {std_total_streams:.2f}")
```

## Categorical Value Analysis

Let's explore the count and uniqueness of categorical values in columns such as 'Sex', 'Country of Origin', 'Primary Language', 'Primary Genre', and 'Artist Type'.

```python
categorical_columns = ['Sex', 'Country of Origin', 'Primary Language', 'Primary Genre', ' Artist Type']

for col in categorical_columns:
    print(f"\n--- Column: {col} ---")

    # Uniqueness (Number of unique values)
    unique_count = df[col].nunique()
    print(f"Number of unique values: {unique_count}")

    # Count (Frequency of each unique value)
    print(f"Value Counts for {col}:")
    print(df[col].value_counts().to_string()) # Print all value counts as string to prevent truncation
```

```python
# Rename multiple columns
# Let's rename 'Primary Language' to 'Language' and 'Primary Genre' to 'Genre'
df_renamed_multiple = df.rename(columns={
    'Primary Genre': 'Genre',
    ' Artist Type': 'Type'
})

print("Original columns:", df.columns.tolist())
print("Columns after renaming multiple columns:", df_renamed_multiple.columns.tolist())
display(df_renamed_multiple.head(2))
```
